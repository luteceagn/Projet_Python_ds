I- Configuration de l'environnement de travail

Chargement des données et des packages, configuration de l'affichage

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

data = pd.read_csv("df_sncf_avec_meteo.csv")
meteo = pd.read_csv("data_meteo_enrichie.csv", sep=';')
# Configuration pour afficher toutes les colonnes et éviter les troncatures
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
pd.set_option('display.float_format', lambda x: '%.2f' % x)

# Style par défaut pour les graphiques
sns.set_theme(style="whitegrid")
```

II- Présentation et analyse de la base de données météorologique

```python
# 1. Vue d'ensemble de la structure et des types
print("Dimensions du DataFrame :", meteo.shape)
display(meteo.info())
# 2. Premières lignes de la base de données
display(meteo.head(5))
# 3. Statistiques descriptives de base (moyennes, min, max, quartiles)
display(meteo.describe())
```

La base de données climatologique a été récupéré via l'API de météo france et ne contient que des données en Open Data. COnstruire cette base a demandé de récupérer les données mensuelles de 2018 à 2026 de toutes les stations météo de France en les récuparants une par une par département

Notre base de données météorologique est un dataframe de 357.361 observations et 158 variables. Une pré séléction des variables ramène ces 158 à une séléction de 13 variables : departement, DATE, NBJGELEE, NBJSOLNG, TN, RR, NBJRR10, NBJORAG, NBJBROU, NBJFF16, TX, NBJTX30, TAMPLIM. Dont 12 réellement explicative car les variables de département et de DATE sont présente pour permettre la jointure avec la base de données SNCF. 

```python
colonnes_a_garder = ['departement','DATE','NBJGELEE','NBJSOLNG','TN','RR','NBJRR10','NBJORAG','NBJBROU','NBJFF16','TX','NBJTX30','TAMPLIM']
meteo = meteo[colonnes_a_garder]
```

```python
# 1. Vue d'ensemble de la structure et des types
print("Dimensions du DataFrame :", data.shape)
display(data.info())
display(meteo.info())
# 2. Premières lignes pour vérifier la fusion
display(data.head(5))
display(meteo.head(5))
# 3. Statistiques descriptives de base (moyennes, min, max, quartiles)
display(data.describe())
display(meteo.describe())

# 5. Premier graphique : Distribution d'une variable clé (ex: la température max au départ)
plt.figure(figsize=(10, 5))
sns.histplot(data['retard_moyen_arrivee'], kde=True, bins=30, color='royalblue')
plt.title("Distribution des retards à l'arrivée")
plt.xlabel('Retard moyen (min)')
plt.ylabel('Nombre d\'observations')
plt.show()
```
